### Messages

messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:
* Role - Identifies the message type (e.g. system, user)
* Content - Represents the actual content of the message (like text, images, audio, documents, etc.)
* Metadata - Optional fields such as response information, message, IDs, and token usages
##### LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
import os 
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=ChatGroq(model="qwen/qwen3-32b")

```

```python
model.invoke("Hello!")
```

### Text Prompts
Text prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history

```python
model.invoke("What is langchain?")
```

Use text prompts when:
* You have single, standalone request
* You don't need conversation history
* You want minimal code complexity

### Message Prompts
Alternatively, you can pass in a list of messages to the model by providing a list of message objects

Messages Types
* System message - Tells the model how to behave and provide context for interactions
* Human message - Represents user input and interactions with the model
* AI message - Responses generated by the model, including text content, tool calls, and metadata
* Took message - Represents the outputs of tool calls

### System Message
A SystemMessage represent an initial set of instructions that primes the models's behavior. You can use a system message to set the tone, define the model's role, and establish guidenlines for responses.


### AI Message
An AIMessage represents the output of a model invocation. Then can include multimodal data, tool calls, and provider-specific metadata that you can later access.


### Tool Message
For models taht support tool calling, AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execution back to the model.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert."),
    HumanMessage("Write a poem on me. My name is Md. Mizanur Rahman. I am learning Machine Learning."),
]

response=model.invoke(messages)
response.content
```

```python
system_msg=SystemMessage("You are a poetry expert.")
messages=[
    system_msg,
    HumanMessage("Write a poem on me. My name is Md. Mizanur Rahman. I am learning Machine Learning."),
]
response=model.invoke(messages)
print(response.content)
```

```python
## Detailed info to the LLM through System message
from langchain.messages import SystemMessage, HumanMessage

system_msg=SystemMessage("""
Your are a helpful assistant for writing poems. You will be given some information about a person, and you will write a poem based on that information. The poem should be creative and engaging, and it should capture the essence of the person based on the information provided.
""")

messages=[
    system_msg,
    HumanMessage("Write a poem on me. My name is Md. Mizanur Rahman. I am learning Machine Learning."),
]
response=model.invoke(messages)
print(response.content)
```

```python
## Message Metadta
human_msg=HumanMessage(
    content="Write a poem on me. My name is Md. Mizanur Rahman. I am learning Machine Learning.",
    name='Mizanur Rahman',
    id="19895"
)

response=model.invoke([
    human_msg
])
response.content
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# Create an AI message manually (e.g., for conversation history). It is not generated by AI.
ai_msg = AIMessage(content="Hello! How can I assist you today?")

# Add to conversation history
messages = [
    SystemMessage(content="You are a helpful assistant."),
    ai_msg,
    HumanMessage(content="Can you tell me a joke?"),
    ai_msg,
    HumanMessage(content="Can you tell me a joke?")
]

response=model.invoke(messages)
print(response.content)
```

#### Tool Message

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes tool call
# (Here, we demostrate manually creating the messages for brevity)
ai_message=AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"New York"},
        "id":"call_123"
    }]
)

#Execute tool and create result message
weather_result= "Sunny, 27°C"
tool_message=ToolMessage(
    content=weather_result,
    tool_call_id="call_123" # Must match the call Id
)

# Continue conversation
messages=[
    ai_message, # Model's tool call
    tool_message, # Tool execution result
    HumanMessage(content="Thanks! What about tomorrow's weather?")
]

response=model.invoke(messages) # Model processes the result.
print(response.content)
```
