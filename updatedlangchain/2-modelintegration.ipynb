### Models Integration With OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1") # I can change the model name to use different models, such as "gpt-3.5-turbo"
model
```

```python
## Invoke the model
response=model.invoke("Hello, How are you?")
response
```

```python
response.content
```

```python
### ChatOpenAI
from langchain_openai import ChatOpenAI
model=ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hello, How are you doing")
response.content
```

### Google Gemini Model Integration

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
model=init_chat_model("google_genai:gemini-2.5-flash-lite")
response=model.invoke("Hello, How are you doing")
response.content
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")

model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response=model.invoke("Hello!")
response.content
```

### GROQ Model Integration

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("Hello")
response.content
```

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=ChatGroq(model="qwen/qwen3-32b")
response=model.invoke("Hello")
response.content
```
